```python
import pandas as pd

data = {
    "mobile": ["A", "B", "C"],
    "battery_size": [4000, None, 5000]
}

df = pd.DataFrame(data)

df = df.dropna(subset=['battery_size'])
df.info()
```

```python
df = pd.read_csv("Mobile_data.csv")
df.isnull().sum()
```

```python
df.isnull().sum()
df.duplicated().sum()
df.drop_duplicates()
```

```python
x = df['highest_price'].mean()
newdf= df.fillna(x)
newdf.info()
newdf.head(10)
```

```python
# df["highest_price"]=df.fillna(["highest_price"]).mean()
df = df.fillna(x)
df.info()
```

```python
df["highest_price"]=df["highest_price"].fillna(
    df.groupby('brand_name')['highest_price'].transform('mean'))
df.info()
df.head(10)
```

```python
# df["highest_price"]=df["highest_price"].fillna(df['highest_price'].mean())
df["lowest_price"]=df["lowest_price"].fillna(df['lowest_price'].mean())

df.info()
# df.head(10)
```

```python
df.dropna(subset=['os'])
# df.head(10)
df.info()
```
