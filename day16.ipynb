# Machine learning : Supervised Learning, unsupervised and reforcement 

Machine Learning Types
Machine Learning is a branch of Artificial Intelligence where computers learn patterns from data and make decisions or predictions without being explicitly programmed.

# There are mainly 3 types of Machine Learning:

1. Supervised Learning
In supervised learning, the machine learns using labeled data.

👉 Labeled data means:

Input data is given

Correct output/answer is also given

The model learns the relationship between input and output.

Example
Input	Output
Study hours	Exam marks
Email text	Spam or Not Spam
The machine predicts outputs for new data.

Types of Supervised Learning
A. Classification
Used when output is categories/classes.

Example:

Spam or Not Spam

Pass or Fail

Cat or Dog

B. Regression
Used when output is numerical values.

Example:

House price prediction

Temperature prediction

Common Algorithms
Linear Regression

Logistic Regression

Decision Tree

Random Forest

Support Vector Machine (SVM)

Real-Life Uses
Email spam detection

Face recognition

Medical diagnosis

Weather forecasting

2. Unsupervised Learning
In unsupervised learning, the machine learns from unlabeled data.

👉 No correct answer/output is provided.

The machine finds hidden patterns or groups by itself.

Example
A shopping website groups customers based on buying behavior.

Types
A. Clustering
Grouping similar data together.

Example:

Customer segmentation

Grouping similar news articles

B. Association
Finding relationships between items.

Example:

“People who buy bread also buy butter.”

Common Algorithms
K-Means Clustering

Hierarchical Clustering

Apriori Algorithm

Real-Life Uses
Recommendation systems

Market analysis

Fraud detection

Social network analysis

3. Reinforcement Learning
In reinforcement learning, an agent learns by interacting with the environment.

The agent:

Takes actions

Receives rewards or penalties

Learns the best action through trial and error

Example
Teaching a robot to walk:

Correct movement → Reward

Wrong movement → Penalty

Main Components
Agent

Environment

Action

Reward

Real-Life Uses
Self-driving cars

Game playing AI

Robotics

Traffic signal control

Simple Comparison Table
Feature	Supervised	Unsupervised	Reinforcement
Data Type	Labeled	Unlabeled	Interaction-based
Goal	Predict output	Find patterns	Maximize reward
Feedback	Correct answer available	No feedback	Reward/Penalty
Example	Spam detection	Customer grouping	Robot learning
Easy Memory Trick
Supervised → Teacher gives answers

Unsupervised → Find patterns yourself

Reinforcement → Learn by rewards and punishment
