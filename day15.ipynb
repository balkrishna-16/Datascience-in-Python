```python
import matplotlib.pyplot as plt
x = [-100, -200, 0, 120, 220]
y = [100, 400, 0, 100, 400]
z = [150, 450, 0, 200, 600]
plt.plot(x, color='red', marker='o', linestyle='--')
#plt.plot(y, color='blue', marker='o', linestyle='--')
plt.plot(z, color='green', marker='o', linestyle='--')
plt.plot(y, marker='D', ms=10, mfc='yellow', mec='red', mew=2)
plt.title("Sample Line Plot")
plt.xlabel("X Axis Label")
plt.ylabel("Y Axis Label")
plt.grid(True)
plt.show()
```
