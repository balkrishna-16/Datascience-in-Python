```python
#Write a program calaculated add, sunstract, mul, div and sqr of ann array in numpy.
#also find the data of each variable taking its mean value. Show data greater than mean value.

import numpy as np

# array
arr = np.array([20, 30, 40, 50])

# operations
a1 = arr + 5
b1 = arr - 10
c1 = arr * 5
d1 = arr / 5
e1 = arr ** 2

# mean
mean = np.mean(arr)

# values greater than mean
greater = arr > mean

print("Add:", a1)
print("Subtract:", b1)
print("Multiply:", c1)
print("Divide:", d1)
print("Square:", e1)
print("Mean:", mean)
print("Greater than mean (True/False):", greater)
print("Values greater than mean:", arr[arr > mean])

```

```python
# Find mean, median, mode, standard deviation, variance, min, max of an array
a = np.array([10,11,12,13,14])
mean= np.mean(a)
median= np.median(a)
# mode = np.
std = np.std(a)
var = np.var(a)
min = np.min(a)
max = np.max(a)
print("mean:",mean)
print("median:",median)
print("standard deviation:",std)
print("variance:",var)
print("min:",min)
print("max:",max)
```
