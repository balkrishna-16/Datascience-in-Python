# Section C: Broadcasting

```python
# Add scalar 10 to matrix.
import numpy as np

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

# add scalar 10
result = matrix + 10

print(result)
```

```python
# Add vector: [10, 20, 30]
# to matrix rows.

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

vector = np.array([10, 20, 30])

# add vector to matrix rows
result = matrix + vector

print(result)
```

```python
# Multiply column vector with matrix

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

# column vector
col_vector = np.array([[10],
                       [20]])

# multiply column vector with matrix
result = matrix * col_vector

print(result)
```

```python
# Broadcast row vector.
import numpy as np

matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

# row vector
row_vector = np.array([10, 20, 30])

# broadcasting row vector
result = matrix + row_vector

print(result)
```

# Explain broadcasting rules.
1. NumPy Broadcasting Rules
Ans: Broadcasting is a NumPy feature that allows arrays of different shapes to perform operations together automatically.
Broadcasting Rules :-

Rule 1: Same Shape
    import numpy as np
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
Output:
[11 22 33]

print(a + b)

Rule 2: One Dimension is 1
    If one dimension is 1, NumPy stretches it automatically.
import numpy as npa = np.array([[1], [2], [3]])b = np.array([10, 20, 30])print(a + b)
Output
[[11 21 31] [12 22 32] [13 23 33]]

Rule 3: Scalar Broadcasting

    A single number (scalar) is applied to all elements.

import numpy as np

a = np.array([1, 2, 3])

print(a + 5)
Output
[6 7 8]

Rule 4: Incompatible Shapes → Error
    import numpy as npa = np.array([1, 2, 3])b = np.array([1, 2])print(a + b)
❌ Error
ValueError: operands could not be broadcast together
Because:


shape (3,)


shape (2,)


These are not compatible.

# Section D: Mathematical Functions

```python
# Dataset:
# sales = np.array([120, 140, 200, 250, 300])
import numpy as np

# dataset
sales = np.array([120, 140, 200, 250, 300])

print(sales)
```

```python
# Calculate: •	sum,  •	mean,  •	median 

total = np.sum(sales)
mean = np.mean(sales)
median = np.median(sales)

print("Sum:", total)
print("Mean:", mean)
print("Median:", median)
```

```python
# Calculate: •	std ,  •	variance 

std = np.std(sales)
variance = np.var(sales)

print("Standard Deviation:", std)
print("Variance:", variance)
```

```python
# Find: •	min ,  •	max 

minimum = np.min(sales)
maximum = np.max(sales)

print("Minimum:", minimum)
print("Maximum:", maximum)
```

```python
numbers = np.array([4, 9, 16, 25, 36])

# calculate square root
result = np.sqrt(numbers)

print(result)
```

```python
# Calculate log values.
result= np.log(numbers)
print(result)
```

```python

numbers = np.array([1, 10, 100, 1000])

# natural log (base e)
ln_values = np.log(numbers)

# log base 10
log10_values = np.log10(numbers)

# log base 2
log2_values = np.log2(numbers)

print("Natural log:", ln_values)
print("Log10:", log10_values)
print("Log2:", log2_values)
```

```python
# Calculate cumulative sum.

sales = np.array([120, 140, 200, 250, 300])
# cumulative sum
cumsum = np.cumsum(sales)

print(cumsum)
```
