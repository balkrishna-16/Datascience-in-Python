```python
import csv
with open("customers-100.csv","r") as f:
    reader =csv.DictReader(f)
    customers = list(reader)

print(f"Total records loaded : {len(customers)}")
print("simple records")
print(customers[0])
```

```python
for c in customers:
    c["Index"]=int(c["Index"])
print("Converted 'index' into 'integer'")
print(type(customers[0]["Index"]))
```

```python
names=[c["First Name"] for c in customers]
print("First 5 Names :",names[:5])
```

```python
names=[c["Country"] for c in customers]
print("First 5 Names :",names[:5])
```

```python
chile_customers = list(filter(lambda c: c["Country"] == "Chile", customers))
print(f"The Chile Customers: {len(chile_customers)}")
print("Sample: ", chile_customers[:2])
```

```python
recent = list(filter(lambda c: c["Subscription Date"] > "2020-12-31", customers))
print(f"The Recent Customers count: {len(recent)}")
```
