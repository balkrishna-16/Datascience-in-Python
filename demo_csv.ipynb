#CSV File Handling

```python
# -------------------------------Writing a CSV file-------------------------------
import csv
header = ["Name", "Age", "City"]
data = [["Balkrishna", 22, "Pampapur"], ["Anshu",21, "Kuiya"], ["Parbin", 20,"Panditpur"]]

with open("brother.csv","w",newline="")as fp:
    wr = csv.writer(fp)
    wr.writerow(header)
    wr.writerows(data)
print("Csv file written successfully!")
```

```python
#-----------------------------------Reading a CSV file---------------------------------
import csv
with open("brother.csv", "r")as fp:
    r=csv.reader(fp)
    for row in r:
        print(row)   #Each row in a list
print("CSV file read Successfully.")
```

```python
# -------------------------------------Append (a):----------------------------------------
import csv

header = ["Name", "Age", "Address"]

data = [
    ["Sudeep", 22, "Pampapur"],
    ["Aashis", 21, "Kuiya"],
    ["Arun", 20, "Panditpur"]
]

with open("brother.csv", "a", newline="") as fp:
    wr = csv.writer(fp)
    wr.writerow(header)
    wr.writerows(data)

print("CSV file written successfully!")
```
