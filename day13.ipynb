```python
import pandas as pd

# Data create
data = {
    "Name": ["Ram", "Shyam", "Hari"],
    "Age": [20, 22, 21],
    "Marks": [80, 75, 90]
}

# DataFrame
df = pd.DataFrame(data)

# Data print
print("Full Data:")
print(df)

# First 2 rows
print("\nFirst 2 rows:")
print(df.head(2))

# Mean (average)
print("\nAverage Marks:")
print(df["Marks"].mean())

# Max marks
print("\nHighest Marks:")
print(df["Marks"].max())
```

```python
# Series
import pandas as pd

a = [1, 7, 2]

myvar = pd.Series(a)

print(myvar)
```

```python
a = [1, 7, 2]

myvar = pd.Series(a, index = ["x", "y", "z"])

print(myvar)
```

```python

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

```python

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index = ["day1", "day2"])

print(myvar)
```

```python
# DataFrame
import pandas as pd

data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

#load data into a DataFrame object:
df = pd.DataFrame(data)

print(df) 
```

```python
#refer to the row index:
print(df.loc[0])
```

```python
#use a list of indexes:
print(df.loc[[0, 1]])
```
