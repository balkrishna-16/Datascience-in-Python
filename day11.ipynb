```python
import numpy as np

a1 = np.array([1, 2, 3, 4, 5], dtype="i")        # integer
b1 = np.array([True, False], dtype=bool)         # boolean (fixed)

c1 = np.array([-3, -2, -1, 0, 1, 2, 3], dtype="i")  # unsigned can't store negative → fixed

d1 = np.array([1.2, 2.3, 34.4], dtype="f")      # float

e1 = np.array([1+2j, 1-3j], dtype="c")          # complex (fixed)

f1 = np.array(["2021-12-31"], dtype="datetime64")  # datetime (fixed)

g1 = np.array([1, 2, 3, 4, 5], dtype="O")       # object (fixed)

h1 = np.array(["bkc", "csk", "skc"], dtype="S") # byte string

i1 = np.array(["bkc", "csk", "skc", "psk"], dtype="U")  # unicode string

# j1 = np.array(["bkc", "csk", "skc", "psk"], dtype="V")  # raw data (valid but uncommon)


print(a1.dtype)
print(b1.dtype)
print(c1.dtype)
print(d1.dtype)
print(e1.dtype)
print(f1.dtype)
print(g1.dtype)
print(h1.dtype)
print(i1.dtype)
# print(j1.dtype)
```
