# Project Work on the topic of "Smart Farming Crop Yield"

```python
import pandas as pd
sf = pd.read_csv("Smart_Farming_Crop_Yield_2024.csv")
sf
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data cleaning

```python
sf = sf[['farm_id', 'crop_type', 'temperature_C', 'soil_moisture_%', 'soil_pH', 'sunlight_hours', 'yield_kg_per_hectare','region', 'crop_disease_status']]
sf.dropna(inplace=True)
sf
```

```python
sf = sf.dropna()
sf
```

```python
plt.figure()
sf['crop_type'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Crop Types')
plt.xlabel('Crop Types')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(sf["yield_kg_per_hectare"], bins=20)
plt.title('Distribution of Crop Yields')
plt.xlabel('Yield (kg/ha)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sf['crop_type'].value_counts().sort_index().plot(kind='line')
plt.title('Crop Types Over Time')
plt.xlabel('Crop Type')
plt.ylabel('Count')
plt.show()  
```

```python
import matplotlib.pyplot as plt
yield_by_crop = sf.groupby('crop_type')['yield_kg_per_hectare'].sum()

plt.figure(figsize=(8,5))
yield_by_crop.plot(kind='bar',color='cyan')
plt.title("Total Yield by Crop Type")
plt.xlabel("Crop Type")
plt.ylabel("Yield (kg/ha)")
plt.show()
```

```python
yield_category = sf['crop_type'].value_counts()

plt.figure(figsize=(8,5))
yield_category.plot(kind='pie', autopct='%1.1f%%')
plt.title("Yield Distribution by Crop Type")
plt.show()
```

```python
item_region = sf.groupby('region')['crop_type'].count().sort_values(ascending=False).head(5)

plt.figure(figsize=(10,5))
item_region.plot(kind='barh', color='teal')
plt.title("Top 5 Regions by Crop Type")
plt.xlabel("Number of Crops")
plt.show()
```
