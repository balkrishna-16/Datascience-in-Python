```python
#------------------ Writing a CSV file -------------------------------
import csv
p = open("student.csv", "w")
swriter = csv.writer(p)
swriter.writerow(['Rollno.','Name','Marks'])
for i in range(3):
    print("Record Number", i+1)
    rno = int(input("Enter Rollno.: "))
    name = input("Enter Name : ")
    marks = float(input("Enter Marks: "))
    srecord = [ rno, name, marks]
    swriter.writerow(srecord)
    p.flush()
p.close()
```

```python
#------------------------------Reading a CSV file ------------------------------
import csv
with open("student.csv","r", newline='\r\n') as fp:
    empdata = csv.reader(fp)
    for r in empdata:
        print(r)
```
