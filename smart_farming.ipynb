```python
import pandas as pd
sf = pd.read_csv("Smart_Farming_Crop_Yield_2024.csv")
sf
```

# Quick look at the data structure

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
print('Data shape:', sf.shape)
print('Columns:', sf.columns.tolist())
```

# Data cleaning

```python
sf = sf[['farm_id', 'crop_type', 'temperature_C', 'soil_moisture_%', 'soil_pH', 'sunlight_hours', 'yield_kg_per_hectare','region', 'crop_disease_status','sensor_id', 'irrigation_type']]
sf.dropna(inplace=True)
sf
```

```python
# Display a summary of the cleaned data
print(sf.info())
```

# 1. Crop Type Analysis


```python
plt.figure()
sf['crop_type'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.title('Top 10 Crop Types')
plt.xlabel('Crop Types')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(sf["yield_kg_per_hectare"], bins=20)
plt.title('Distribution of Crop Yields')
plt.xlabel('Yield (kg/ha)')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7,5))
sf['crop_type'].value_counts().sort_index().plot(kind='line')
plt.title('Crop Types Over Time')
plt.xlabel('Crop Type')
plt.ylabel('Count')
plt.show()  
```

```python
import matplotlib.pyplot as plt
yield_by_crop = sf.groupby('crop_type')['yield_kg_per_hectare'].sum()

plt.figure(figsize=(8,5))
yield_by_crop.plot(kind='bar',color='cyan')
plt.title("Total Yield by Crop Type")
plt.xlabel("Crop Type")
plt.ylabel("Yield (kg/ha)")
plt.show()
```

# Pie Chart - Crop Distribution


```python
yield_category = sf['crop_type'].value_counts()

plt.figure(figsize=(10,7))
yield_category.plot(kind='pie', autopct='%1.1f%%')
plt.title("Yield Distribution by Crop Type")
plt.show()
```

# Horizontal Bar Chart - Region Wise Farm Count

```python
item_region = sf.groupby('region')['crop_type'].count().sort_values(ascending=True).head(5)

plt.figure(figsize=(10,5))
item_region.plot(kind='barh', color='teal')
plt.title("Top 5 Regions by Crop Type")
plt.xlabel("Number of Crops")
plt.show()
```

```python
## Histogram of Yield
plt.figure(figsize=(8,6))
sns.histplot(sf['yield_kg_per_hectare'], kde=True, bins=30, color='skyblue')
plt.xlabel('Yield (kg per hectare)')
plt.title('Distribution of Crop Yield')
plt.show()
```

```python
# 1. Crop Type Analysis
# -------------------------------

# Bar Chart - Crop Count
plt.figure(figsize=(8,5))
sns.countplot(x='crop_type', data=sf)
plt.title("Crop Type Count")
plt.xticks(rotation=45)
plt.show()
```

```python
# Pie Chart - Crop Distribution
crop_counts = sf['crop_type'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(crop_counts, labels=crop_counts.index, autopct='%1.1f%%')
plt.title("Crop Type Distribution")
plt.show()
```

```python
# Horizontal Bar Chart - Region Wise Farm Count
plt.figure(figsize=(8,5))
sf['region'].value_counts().plot(kind='barh')
plt.title("Region Wise Farm Count")
plt.xlabel("Count")
plt.show()
```

```python
# 2. Yield Analysis
# -------------------------------

# Top 10 Farms by Yield
top_farms = sf.groupby('farm_id')['yield_kg_per_hectare'].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_farms.plot(kind='bar')
plt.title("Top 10 Farms by Yield")
plt.ylabel("Yield (kg/hectare)")
plt.xticks(rotation=45)
plt.show()
```

```python
# Histogram - Yield Distribution
plt.figure(figsize=(8,5))
sns.histplot(sf['yield_kg_per_hectare'], bins=20, kde=True)
plt.title("Yield Distribution")
plt.xlabel("Yield")
plt.show()
```

```python
# Box Plot - Yield Range
plt.figure(figsize=(8,5))
sns.boxplot(x=sf['yield_kg_per_hectare'])
plt.title("Yield Box Plot")
plt.show()
```

```python
# Most Used Irrigation Type
plt.figure(figsize=(8,5))
sns.countplot(x='irrigation_type', data=sf)
plt.title("Most Used Irrigation Method")
plt.xticks(rotation=45)
plt.show()
```

```python
# Highest Average Yield by Irrigation Type
avg_yield = sf.groupby('irrigation_type')['yield_kg_per_hectare'].mean().sort_values(ascending=False)
plt.figure(figsize=(8,5))
avg_yield.plot(kind='bar', color='green')
plt.title("Average Yield by Irrigation Type")
plt.ylabel("Average Yield")
plt.xticks(rotation=45)
plt.show()
```

```python
# 4. Disease Analysis
# -------------------------------

# Disease Status Count
plt.figure(figsize=(7,5))
sns.countplot(x='crop_disease_status', data=sf)
plt.title("Crop Disease Status")
plt.show()
```

```python
# Pie Chart - Disease Percentage
disease_counts = sf['crop_disease_status'].value_counts()

plt.figure(figsize=(7,7))
plt.pie(disease_counts, labels=disease_counts.index, autopct='%1.1f%%')
plt.title("Disease Status Percentage")
plt.show()
```

```python
# Box Plot - Yield vs Disease Status
plt.figure(figsize=(8,5))
sns.boxplot(x='crop_disease_status', y='yield_kg_per_hectare', data=sf)
plt.title("Yield Comparison by Disease Status")
plt.show()
```

```python
# 5. Environmental Factor Analysis
# -------------------------------

# Temperature vs Yield
plt.figure(figsize=(8,5))
sns.scatterplot(x='temperature_C', y='yield_kg_per_hectare', data=sf, color='red')
plt.title("Temperature vs Yield")
plt.show()
```

```python
# Soil Moisture vs Yield
plt.figure(figsize=(8,5))
sns.scatterplot(x='soil_moisture_%', y='yield_kg_per_hectare', data=sf)
plt.title("Soil Moisture vs Yield")
plt.show()
```

```python
# Sunlight vs Yield
plt.figure(figsize=(8,5))
sns.scatterplot(x='sunlight_hours', y='yield_kg_per_hectare', data=sf, color='orange')
plt.title("Sunlight Hours vs Yield")
plt.show()
```

```python
# Final Insights
# -------------------------------

print("\n========== PROJECT INSIGHTS ==========")

print("\nMost Common Crop Type:")
print(sf['crop_type'].mode()[0])

print("\nBest Irrigation Type Based on Yield:")
print(avg_yield.idxmax())

print("\nAverage Yield:")
print(sf['yield_kg_per_hectare'].mean())

print("\nRegion with Highest Average Yield:")
print(sf.groupby('region')['yield_kg_per_hectare'].mean().idxmax())

print("\n======================================")
```
